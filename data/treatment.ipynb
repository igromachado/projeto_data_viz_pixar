```python
import pandas as pd
```

```python
# reading every dataframe
df_academy = pd.read_csv('academy.csv')
df_box = pd.read_csv('box_office.csv')
df_genre = pd.read_csv('genres.csv')
df_films = pd.read_csv('pixar_films.csv')
df_rating = pd.read_csv('public_response.csv')

dfs = {
    "pixar_films.csv": df_films,
    "public_response.csv": df_rating,
    "box_office.csv": df_box,
    "genres.csv": df_genre,
    "academy.csv": df_academy
}
```

```python
# dropping unecessary columns
df_films.drop(columns=['number'], inplace=True)
```

```python
# inspecting the values of the dataframes
for name, df in dfs.items():
    print(name)
    print('shape', df.shape)
    print(df.info())
    print()
```

```python
# checking null values
for name, df in dfs.items():
    print(df)
```

```python
# checking null values
for name, df in dfs.items():
    print(name)
    print(df.isna().sum())
    print()
```

```python
df_box.loc[df_box['film'] == 'Luca', 'budget'] = 135000000
```

```python
# checking null values
for name, df in dfs.items():
    print(name, df.duplicated().sum())
    
    if df.duplicated().sum() > 0:
        dfs[name] = df.drop_duplicates()
```

```python
# checking dtypes
for name, df in dfs.items():
    print(name)
    print(df.dtypes)
    print()
```

```python
df_films['release_date'] = pd.to_datetime(df_films['release_date'], errors='coerce')
```

```python
for df, name in dfs.items():
    df.to_csv(name, index=False)
```
